```python
# Import necessary libraries for data manipulation and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Car Dekho dataset into a pandas DataFrame
df=pd.read_csv("/content/1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")
```

```python
# Display the first 5 rows of the dataset to understand the structure
df.head()
```

```python
# Check the dimensions of the dataset (rows, columns)
df.shape
```

```python
# Identify missing values in each column
df.isnull().sum()
```

```python
# View summary information including data types and non-null counts
df.info()
```

```python
# Generate descriptive statistics for numerical columns
df.describe()
```

```python
# Plot relationship between Present Price and Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Present_Price", y="Selling_Price")
plt.title("Present Price vs Selling Price")
plt.xlabel("Present Price")
plt.ylabel("Selling Price")
plt.show()
```

### Inferences: Present Price vs Selling Price
* There is a strong positive correlation between the `Present_Price` and `Selling_Price`.
* As the showroom price (Present Price) of a car increases, its resale value (Selling Price) generally increases as well.
* Most cars are concentrated in the lower price range (0-20 lakhs).

```python
# Plot relationship between Kilometers Driven and Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Kms_Driven", y="Selling_Price")
plt.title("Kilometers Driven vs Selling Price")
plt.show()
```

### Inferences: Kilometers Driven vs Selling Price
* Generally, cars with lower mileage (`Kms_Driven`) tend to fetch a higher `Selling_Price`.
* However, the relationship is less linear than price, suggesting that other factors like brand and condition also play significant roles.

```python
# Plot relationship between Manufacturing Year and Selling Price
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Year", y="Selling_Price")
plt.title("Manufacturing Year vs Selling Price")
plt.show()
```

### Inferences: Manufacturing Year vs Selling Price
* Newer models (higher `Year`) consistently have higher selling prices compared to older ones.
* This highlights the impact of depreciation over time on the resale value of vehicles.

```python
# Compare Selling Price distributions across different Fuel Types
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Fuel_Type", y="Selling_Price")
plt.title("Selling Price by Fuel Type")
plt.show()
```

### Inferences: Selling Price by Fuel Type
* **Diesel** cars tend to have a significantly higher average selling price compared to **Petrol** and **CNG** cars.
* This might be due to higher initial costs and better fuel efficiency/torque preferred in the resale market for larger vehicles.

```python
# Compare Selling Price distributions across different Transmission types
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Transmission", y="Selling_Price")
plt.title("Selling Price by Transmission")
plt.show()
```

### Inferences: Selling Price by Transmission
* **Automatic** transmission cars have a higher median and average selling price than **Manual** cars.
* This reflects the premium nature and growing demand for automatic vehicles.

```python
# Compare Selling Price distributions across different Seller Types
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Seller_Type", y="Selling_Price")
plt.title("Selling Price by Seller Type")
plt.show()
```

### Inferences: Selling Price by Seller Type
* Cars sold by **Dealers** command much higher prices than those sold by **Individuals**.
* This could be because dealers often refurbish cars, offer warranties, or deal with more premium models.

```python
# Compare Selling Price distributions based on the number of previous owners
plt.figure(figsize=(7,5))
sns.boxplot(data=df, x="Owner", y="Selling_Price")
plt.title("Selling Price by Number of Previous Owners")
plt.show()
```

### Inferences: Selling Price by Number of Previous Owners
* Cars with **0 previous owners** (first-hand resale) have the highest selling price.
* As the number of owners increases, the resale value drops significantly, likely due to concerns over vehicle wear and maintenance history.

```python
# Create a heatmap to visualize correlations between numerical features
plt.figure(figsize=(9,6))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
            annot=True,
            cmap="coolwarm",
            fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

### Inferences: Correlation Heatmap
* **Highest Correlation:** `Present_Price` and `Selling_Price` (0.88), confirming they move together.
* **Negative Correlation:** `Year` and `Kms_Driven` (-0.52), which is logical as older cars usually have more mileage.
* `Selling_Price` also shows a decent positive correlation with `Year` (0.24), indicating newer cars are more valuable.

```python
# Calculate and print averages for key metrics and categorical groupings
print("Average Selling Price:", df["Selling_Price"].mean())
print("Average Present Price:", df["Present_Price"].mean())
print("Average Kms Driven:", df["Kms_Driven"].mean())

# Grouped averages to identify trends
print(df.groupby("Fuel_Type")["Selling_Price"].mean())
print(df.groupby("Transmission")["Selling_Price"].mean())
print(df.groupby("Seller_Type")["Selling_Price"].mean())
```
